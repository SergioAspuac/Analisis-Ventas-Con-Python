```python
import pandas as pd
import matplotlib.pyplot as plt

# Datos de ejemplo
data = {
    "Ciudad": ["Guatemala", "Monterrey", "Guatemala", "Antigua"],
    "Producto": ["Laptop", "Tablet", "Telefono", "Laptop"],
    "Ventas": [2400, 3000, 1500, 2000]
}

# Crear DataFrame
df = pd.DataFrame(data)

# Agrupar por ciudad
ventas_ciudad = df.groupby("Ciudad")["Ventas"].sum()

# Mostrar resultados
print(ventas_ciudad)

# Gráfica
ventas_ciudad.plot(kind="bar", title="Ventas por Ciudad")
plt.xlabel("Ciudad")
plt.ylabel("Ventas")
plt.show()
```
